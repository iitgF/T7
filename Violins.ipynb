```r
library(readxl)
library(ggplot2)
library(dplyr)
library(gridExtra)

# Load the data from the Excel file
file_path <- "Nov_to_Apr_violins.xlsx"
data <- read_excel(file_path)

# Select relevant columns and remove rows with missing data
selected_data <- data %>%
  select(Month_End, PnL_Distribution) %>%
  filter(!is.na(Month_End) & !is.na(PnL_Distribution))

# Convert Month_End to factor to order the months explicitly
month_levels <- unique(selected_data$Month_End)
selected_data$Month_End <- factor(selected_data$Month_End, levels = month_levels)

# Create individual violin plots for each Month_End
violin_plots <- lapply(month_levels, function(month) {
  ggplot(filter(selected_data, Month_End == month), aes(x = factor(1), y = PnL_Distribution)) +
    geom_violin(trim = FALSE, fill = "lightblue", color = "darkblue") +
    labs(title = as.character(month), x = NULL, y = "PnL Distribution") +
    theme_minimal() +
    theme(axis.text.x = element_blank(), axis.ticks.x = element_blank())
})

# Arrange the six violin plots in a grid (2 rows x 3 columns)
grid.arrange(grobs = violin_plots, ncol = 3)
```
